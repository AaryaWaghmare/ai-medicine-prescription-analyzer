```python
import gradio as gr
import cv2
import pytesseract
import re
from difflib import SequenceMatcher

def analyze_medicine(image_path):

    # Read image
    img = cv2.imread(image_path)

    # Convert to grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Resize image
    gray = cv2.resize(gray, None, fx=2, fy=2)

    # Improve contrast
    processed = cv2.equalizeHist(gray)

    # OCR
    extracted_text = pytesseract.image_to_string(
        processed,
        config="--psm 6"
    )

    # Convert OCR text to lowercase
    ocr_text = extracted_text.lower()

    # Extract words
    ocr_words = re.findall(r'[a-zA-Z]+', ocr_text)

    detected = []

    # Fuzzy matching
    for medicine in medicine_df["Medicine"]:
        medicine_words = medicine.lower().split()

        for med_word in medicine_words:
            for word in ocr_words:

                similarity = SequenceMatcher(
                    None,
                    med_word,
                    word
                ).ratio()

                if similarity >= 0.70:
                    detected.append(medicine)
                    break

            if medicine in detected:
                break

    # Remove duplicates
    detected = list(dict.fromkeys(detected))

    # Create result
    if detected:

        result_text = "AI MEDICINE & PRESCRIPTION ANALYZER\n"
        result_text += "=" * 45 + "\n\n"

        for medicine in detected:
            result = medicine_df[
                medicine_df["Medicine"] == medicine
            ].iloc[0]

            result_text += f"Medicine Detected : {result['Medicine']}\n"
            result_text += f"Strength          : {result['Strength']}\n"
            result_text += f"Form              : {result['Form']}\n"
            result_text += f"Category          : {result['Category']}\n\n"

        result_text += "-" * 45 + "\n"
        result_text += "Educational prototype only.\n"
        result_text += "Not a medical diagnosis or prescription."

    else:
        result_text = "No medicine could be identified."

    return result_text


# Create interface
interface = gr.Interface(
    fn=analyze_medicine,
    inputs=gr.Image(
        type="filepath",
        label="Upload Medicine / Prescription Image"
    ),
    outputs=gr.Textbox(
        label="Analysis Result",
        lines=10
    ),
    title="AI Medicine & Prescription Analyzer",
    description="Upload an image to extract and analyze medicine information."
)

interface.launch(share=True)
```

```python
def analyze_medicine_v2(image_path):

    # Read image
    img = cv2.imread(image_path)

    # Convert to grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Resize image
    gray = cv2.resize(gray, None, fx=2, fy=2)

    # Improve contrast
    processed = cv2.equalizeHist(gray)

    # OCR
    extracted_text = pytesseract.image_to_string(
        processed,
        config="--psm 6"
    )

    ocr_text = extracted_text.lower()
    ocr_words = re.findall(r'[a-zA-Z]+', ocr_text)

    detected_brands = []

    # Match OCR words with brand names
    for brand in medicine_db["Brand Name"]:

        similarity = 0

        for word in ocr_words:
            current_similarity = SequenceMatcher(
                None,
                brand.lower(),
                word
            ).ratio()

            similarity = max(similarity, current_similarity)

        if similarity >= 0.70:
            detected_brands.append(brand)

    detected_brands = list(dict.fromkeys(detected_brands))

    # Display result
    if detected_brands:

        result_text = "AI MEDICINE & PRESCRIPTION ANALYZER\n"
        result_text += "=" * 45 + "\n\n"

        for brand in detected_brands:

            result = medicine_db[
                medicine_db["Brand Name"] == brand
            ].iloc[0]

            result_text += f"Brand Name        : {result['Brand Name']}\n"
            result_text += f"Generic Medicine  : {result['Generic Medicine']}\n"
            result_text += f"Strength          : {result['Strength']}\n"
            result_text += f"Form              : {result['Form']}\n"
            result_text += f"Category          : {result['Category']}\n"

        result_text += "\n" + "-" * 45 + "\n"
        result_text += "Educational prototype only.\n"
        result_text += "Not a medical diagnosis or prescription."

    else:
        result_text = "No medicine could be identified."

    return result_text
```

```python
def analyze_medicine_v3(image_path):

    # Read image
    img = cv2.imread(image_path)

    # Preprocessing
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    gray = cv2.resize(gray, None, fx=2, fy=2)
    processed = cv2.equalizeHist(gray)

    # OCR
    extracted_text = pytesseract.image_to_string(
        processed,
        config="--psm 6"
    )

    ocr_text = extracted_text.lower()
    ocr_words = re.findall(r'[a-zA-Z]+', ocr_text)

    best_match = None
    best_score = 0

    # Check both Brand Name and Generic Medicine
    for _, row in medicine_db.iterrows():

        names = [
            row["Brand Name"],
            row["Generic Medicine"]
        ]

        for name in names:

            for word in ocr_words:

                similarity = SequenceMatcher(
                    None,
                    name.lower(),
                    word
                ).ratio()

                if similarity > best_score:
                    best_score = similarity
                    best_match = row

    # Display result
    if best_match is not None and best_score >= 0.70:

        result_text = "AI MEDICINE & PRESCRIPTION ANALYZER\n"
        result_text += "=" * 45 + "\n\n"

        result_text += f"Brand Name        : {best_match['Brand Name']}\n"
        result_text += f"Generic Medicine  : {best_match['Generic Medicine']}\n"
        result_text += f"Strength          : {best_match['Strength']}\n"
        result_text += f"Form              : {best_match['Form']}\n"
        result_text += f"Category          : {best_match['Category']}\n"

        result_text += "\n" + "-" * 45 + "\n"
        result_text += "Educational prototype only.\n"
        result_text += "Not a medical diagnosis or prescription."

    else:
        result_text = "No medicine could be identified."

    return result_text
```

```python
def analyze_medicine_final(image_path):

    # Read image
    img = cv2.imread(image_path)

    # Preprocessing
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    gray = cv2.resize(gray, None, fx=2, fy=2)
    processed = cv2.equalizeHist(gray)

    # OCR
    extracted_text = pytesseract.image_to_string(
        processed,
        config="--psm 6"
    )

    ocr_text = extracted_text.lower()
    ocr_words = re.findall(r'[a-zA-Z]+', ocr_text)

    best_match = None
    best_score = 0

    # Match both brand and generic names
    for _, row in medicine_db.iterrows():

        names = [
            row["Brand Name"],
            row["Generic Medicine"]
        ]

        for name in names:

            for word in ocr_words:

                similarity = SequenceMatcher(
                    None,
                    name.lower(),
                    word
                ).ratio()

                if similarity > best_score:
                    best_score = similarity
                    best_match = row

    # Create result
    if best_match is not None and best_score >= 0.70:

        result_text = "AI MEDICINE & PRESCRIPTION ANALYZER\n"
        result_text += "=" * 45 + "\n\n"

        result_text += "DETECTED MEDICINE\n"
        result_text += "-" * 25 + "\n"
        result_text += f"Brand Name       : {best_match['Brand Name']}\n"
        result_text += f"Generic Medicine : {best_match['Generic Medicine']}\n"
        result_text += f"Strength         : {best_match['Strength']}\n"
        result_text += f"Form             : {best_match['Form']}\n"
        result_text += f"Category         : {best_match['Category']}\n"

        result_text += "\n\nEXTRACTED OCR TEXT\n"
        result_text += "-" * 25 + "\n"
        result_text += extracted_text[:1000]

        result_text += "\n\n" + "-" * 45 + "\n"
        result_text += "Educational prototype only.\n"
        result_text += "Not a medical diagnosis or prescription."

    else:
        result_text = (
            "No recognized medicine was found.\n\n"
            "The image may contain unclear text or a medicine "
            "that is not present in the database."
        )

    return result_text
```

```python
def analyze_medicine_confidence(image_path):

    # Read image
    img = cv2.imread(image_path)

    # Preprocessing
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    gray = cv2.resize(gray, None, fx=2, fy=2)
    processed = cv2.equalizeHist(gray)

    # OCR
    extracted_text = pytesseract.image_to_string(
        processed,
        config="--psm 6"
    )

    ocr_text = extracted_text.lower()
    ocr_words = re.findall(r'[a-zA-Z]+', ocr_text)

    best_match = None
    best_score = 0

    # Compare OCR words with brand and generic names
    for _, row in medicine_db.iterrows():

        names = [
            row["Brand Name"],
            row["Generic Medicine"]
        ]

        for name in names:

            for word in ocr_words:

                similarity = SequenceMatcher(
                    None,
                    name.lower(),
                    word
                ).ratio()

                if similarity > best_score:
                    best_score = similarity
                    best_match = row

    if best_match is not None and best_score >= 0.70:

        confidence = round(best_score * 100, 1)

        result_text = "AI MEDICINE & PRESCRIPTION ANALYZER\n"
        result_text += "=" * 45 + "\n\n"

        result_text += f"Brand Name        : {best_match['Brand Name']}\n"
        result_text += f"Generic Medicine  : {best_match['Generic Medicine']}\n"
        result_text += f"Strength          : {best_match['Strength']}\n"
        result_text += f"Form              : {best_match['Form']}\n"
        result_text += f"Category          : {best_match['Category']}\n"
        result_text += f"Match Confidence  : {confidence}%\n"

        result_text += "\n" + "-" * 45 + "\n"
        result_text += "Educational prototype only.\n"
        result_text += "Not a medical diagnosis or prescription."

    else:
        result_text = (
            "No recognized medicine was found.\n\n"
            "The image may contain unclear text or a medicine "
            "that is not present in the database."
        )

    return result_text
```
